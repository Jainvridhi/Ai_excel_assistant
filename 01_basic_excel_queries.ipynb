```python
file_path = "sales_data.xlsx" 
df = pd.read_excel(file_path)


print("✅ Dataset Loaded Successfully!")
print("Data Preview:")
print(df.head())
```

```python
#Simple Query-to-Code Mapping
def query_to_code(query, df):
    query = query.lower()
    if "average" in query and "sales" in query:
        if "region" in query:
            return df.groupby("Region")["Sales"].mean()
        elif "product" in query:
            return df.groupby("Product")["Sales"].mean()
        else:
            return df["Sales"].mean()
    elif "sum" in query and "sales" in query:
        return df["Sales"].sum()
    elif "total profit" in query:
        return df["Profit"].sum()
    elif "profit by product" in query:
        return df.groupby("Product")["Profit"].sum()
    elif "count" in query:
        return df.shape[0]
    else:
        return "⚠️ Query not understood yet."
```

```python
# Example Queries
# ----------------------------
queries = [
"Find average sales per region",
"Show total profit",
"Sum of sales",
"Profit by product",
"Count rows"
]


for q in queries:
    print("\n Query:", q)
    result = query_to_code(q, df)
    print("Result:\n", result)



```

```python
# Visualization
if isinstance(result, pd.Series):
    result.plot(kind="bar", title=q.title(), color="skyblue")
    plt.ylabel("Value")
    plt.show()
```

```python
# Export results to a new Excel file

def export_to_excel(df, filename="query_results.xlsx"):
    """
    Save the given DataFrame to an Excel file inside the 'data/' folder.
    """
    output_path = f"../data/{filename}"
    df.to_excel(output_path, index=False)
    print(f" Results exported successfully to {output_path}")
```
